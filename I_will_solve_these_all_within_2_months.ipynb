```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# --- Edit these two lines per problem ---
def f(t, s):
    x, y = s
    return [y, -np.sin(x) - 0.3*y]   # example: damped pendulum

xlim, ylim = (-2*np.pi, 2*np.pi), (-3, 3)

# Vector field
X, Y = np.meshgrid(np.linspace(*xlim, 25), np.linspace(*ylim, 25))
U, V = f(0, [X, Y])
plt.figure(figsize=(8, 5))
plt.streamplot(X, Y, U, V, color='gray', density=1.2, linewidth=0.7)

# Sample trajectories from a few initial conditions
for x0 in [(-5, 1.5), (2, 2), (0.5, -1), (4, 0.5)]:
    sol = solve_ivp(f, [0, 40], x0, dense_output=True, rtol=1e-8)
    t = np.linspace(0, 40, 2000)
    plt.plot(*sol.sol(t), lw=1.5)

plt.xlabel('x'); plt.ylabel('y'); plt.title('Phase portrait')
plt.xlim(xlim); plt.ylim(ylim); plt.tight_layout(); plt.show()
```
